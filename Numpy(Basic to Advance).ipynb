BROADCASTING

1.) broadcasting is used on arrays which are not similar in size and the smaller is broadcast means smaller array will be assumed like it is the same size of the bigger array and operations will be implemented

```python
import numpy as np
a = np.arange(1,10).reshape(3,3)
b = np.array([10])

print(a+b)
```

RULES OF BROADCASTING


1.   Make the Two Array of the same dimension
2.   Make each dimension of the two array the same size



```python
#more example
g = np.arange(1,10).reshape(3,3)
f = np.array([10,20,30])

print(g+f)
```

Broadcasting is used in vectorization in which we deal with the huge data and numpy make the calculation super fast

MEAN SQUARED ERROR

```python
actual = np.random.randint(1,50,25)
predicted = np.random.randint(1,50,25)
```

```python
def mase(actual,predicted):
  return np.mean((actual-predicted)**2)

mase(actual,predicted)
```

```python
#binary cross entropy
(actual-predicted)**2
```

Categorical Cross-Entropy is widely used as a loss function to measure how well a model predicts the correct class in multi-class classification problems. It measures the difference between the predicted probability distribution and the true one-hot encoded labels, guiding the model to assign higher probabilities to the correct class.
